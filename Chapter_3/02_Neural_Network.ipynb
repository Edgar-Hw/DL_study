### 3.2 신경망

#### 3.2.1 퍼셉트론

- 신경망은 퍼셉트론(Perceptron) 알고리즘에서부터 시작
- 퍼셉트론 알고리즘은 여러 개의 신호를 받아 하나의 신호를 출력

- $x_1$과 $x_2$는 입력(= 데이터의 특성)
- $y$는 출력(= 정답)
- $w$는 가중치
- 퍼셉트론은 x와 가중치 w를 곱한 값을 모두 더하여 하나의 값(y)으로 만들고 이와 임계값(threshold)과 비교하여
  - 임계값보다 크면 **1** 출력
  - 그렇지 않으면 **0** 출력


$$y=\begin{cases}
0 & (w_1x_1+w_2x_2 \le threshold) \\
1 & (w_1x_1+w_2x_2 > threshold)
\end{cases}$$

- 이렇게 신경망에서 만들어진 값을 적절한 출력값으로 변환해주는 함수를 **활성화 함수**(Activation Function)
- 또한, 여기까지의 과정을 퍼셉트론의 기본 단위라고 한다.

#### OR 게이트 구현해보기

```python
import tensorflow as tf
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import SGD
from tensorflow.keras.losses import mse
```

```python
tf.random.set_seed(777)
```

```python
# 데이터 준비하기
x = np.array([[0, 0], [1, 0], [0, 1], [1, 1]])
y = np.array([[0], [1], [1], [1]])

# 모델 구성하기
model = Sequential()
# 단층 퍼셉트론을 구성
# model.add(Dense(1, input_shape = (2, ), activation='linear'))
# ---- 권장 방식 ----
model.add(Input(shape=(2,)))
model.add(Dense(1, activation='linear'))

# 모델 준비하기
model.compile(optimizer = SGD(), loss = mse, metrics = ['acc'])

# 학습시키기
model.fit(x, y, epochs=500)
```

```python
model.get_weights()
```
