```python
import tensorflow as tf
import numpy as np
```

```python
# 텐서의 차원과 기본 연산
a = tf.constant(2) # 텐서 선언
print(tf.rank(a))  # 해당 텐서의 랭크를 계산
```

```python
b = tf.constant([1, 2])
print(tf.rank(b))
```

```python
c = tf.constant([[1, 2], [3, 4]])
print(tf.rank(c))
```

▶ [요약]
- 출력의 결과는 텐서 형태로 나옴
- 첫 번째 인자는 해당 텐서의 랭크(rank)
- 두 번째 인자는 해당 텐서의 형태(랭크는 스칼라 형태를 나타내기 때문에 shape=()처럼 빈 칸)
- 세 번째 인자는 해당 텐서의 데이터 타입

```python
# 즉시 실행 모드를 통한 연산
a = tf.constant(3)
b = tf.constant(2)

# 기본 연산
# 텐서 형태로 출력해보기
print(tf.add(a, b))
print(tf.subtract(a, b))

# 넘파이 배열 형태로 출력해보기
print(tf.multiply(a, b).numpy())
print(tf.divide(a, b).numpy())
```

```python
# 텐서에서 넘파이로, 넘파이에서 텐서로
c = tf.add(a, b).numpy() # a와 b를 더한 후 NumPy 배열 형태로 변환
c_square = np.square(c, dtype=np.float32) # NumPy 모듈에 있는 square 함수를 적용
c_tensor = tf.convert_to_tensor(c_square) # 다시 텐서로 변환

print(f'numpy array : {c:.1f}, applying square with numpy : {c_square:.1f}, convert_to_tensor : {c_tensor:.1f}')
```

```python
# @tf.function
@tf.function
def square_pos(x):
    if x > 0:
        x = x * x
    else:
        x = x * -1
    return x

print(square_pos(tf.constant(2)))
```
